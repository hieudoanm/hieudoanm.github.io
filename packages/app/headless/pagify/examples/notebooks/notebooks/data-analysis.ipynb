# Data Analysis with pandas

This notebook demonstrates basic data analysis using pandas.

```python
import pandas as pd
import numpy as np

# Create sample data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}
df = pd.DataFrame(data)
df
```

## Data Exploration

Let's explore the dataset structure and basic statistics.

```python
# Display basic information
print(df.info())
print("\n")
print(df.describe())
```

## Filtering Data

Filter the dataset to find people older than 28.

```python
older_than_28 = df[df['Age'] > 28]
older_than_28
```
